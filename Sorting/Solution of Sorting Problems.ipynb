```python
#1
def bubblesortas(lis):
    for j in range(len(lis)):
        for i in range(1,len(lis)):
            if lis[i]<lis[i-1]:
                lis[i],lis[i-1]=lis[i-1],lis[i]
    
    return lis


print('Ascending Order')
print(bubblesortas([10, 1, 20, 3, 6, 2, 5, 11, 15, 2, 12, 14, 17, 18, 29]))
```

```python
#2
def selectionsort(listt):
    for i in range(len(listt)):
        champ = listt[i]
        var = i
        for j in range(i+1,len(listt)):
            if listt[j]<champ:
                champ = listt[j]
                var = j

        listt[i],listt[var] = champ,listt[i]
    print(listt)

selectionsort([10, 1, 20, 3, 6, 2, 5, 11, 15, 2, 12, 14, 17, 18, 29])           
```

```python
#4
def sortinownway(listt):
    flag = True
    for i in range(len(listt)):
        champ = listt[i]
        var = i
        if flag==True:
            for j in range(i+1,len(listt)):
                if listt[j]<champ:
                    champ = listt[j]
                    var = j

            listt[i],listt[var] = champ,listt[i]
            flag = False
        else:
            for j in range(i+1,len(listt)):
                if listt[j]>champ:
                    champ = listt[j]
                    var = j

            listt[i],listt[var] = champ,listt[i]
            flag = True

    print(listt)

sortinownway( [10,30,20,70,11,15,22,16,58,100,12,56,70,80])
```

```python
#5
lst = [ ["Alan", 95, 87, 91], ["Turing", 92, 90, 83], ["Elon", 87, 92, 80], ["Musk", 85, 94, 90] ]
val = input('Subject name: ')
dic = {}
if val == 'MAT110':
    for i in range(len(lst)):
        dic[lst[i][3]]=lst[i][0]
    
elif val=='PHY111':
    for i in range(len(lst)):
        dic[lst[i][2]]=lst[i][0]

elif val=='CSE110':
    for i in range(len(lst)):
        dic[lst[i][1]]=lst[i][0]

dictd = sorted(dic, reverse= True)

for i in dictd:
    print(dic[i])

```
