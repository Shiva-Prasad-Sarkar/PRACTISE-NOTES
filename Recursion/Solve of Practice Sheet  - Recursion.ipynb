```python
#p1
def sumDigits(n):
    if n%10==n:
        return n
    return (n%10) + sumDigits(n//10)


print(sumDigits(12376))
```

```python
def compare(l,r,m):
    if l>m:
        print('lss')
    else:
        print('r')
def do_sum(a):
    if len(a)==0:
        return 0
    else:
        return a[0]+do_sum(a[1::])
    
def g(a):
    s = len(a)
    m = a[s//2]
    l = do_sum(a[0:s//2])
    r = do_sum(a[s//2+1::])
    compare(l,r,m)

g([4,1,6,3,7])
```

```python
#p2
def bunnyEars(n):
    if n==0:
        return 0
    elif n==1:
        return 2
    if n%2==0:
        return 3+bunnyEars(n-1)
    else:
        return 2+bunnyEars(n-1)
    

print(bunnyEars(0))
```

```python
#p3
def count7(n):
    
    if n%10==n:
        if n==7:
         return 1
        else:
           return 0
        
    if (n%10)%7==0:
       return 1+count7(n//10)
    else:
       return count7(n//10)
    
print(count7(223))
```

```python
#ps5
def pirep(s):
    if len(s)<=2:
        if s=='pi':
            return '3.14'
        else:
            return s
    
    if s[0]=='p':
        if s[1]=='i': 
            return '3.14'+pirep(s[2:])
        else:
            return s[0]+pirep(s[1:])

    else:
        return s[0]+pirep(s[1:])
    
print(pirep('kdpippip'))
```

```python
#p6
def array11(arr,ind):
    
    if ind == len(arr)-1:
        if arr[ind] == 11:
            return 1
        else:
            return 0
    else:
        if arr[ind]==11:
            return 1+array11(arr,ind+1)
        else:
            return array11(arr,ind+1)
    
print(array11([1, 2,], 0))
```

```python
#p7
def pairStar(s):
    if len(s)==1:
        return s
    
    if s[0]==s[1]:
        return s[0]+'*'+pairStar(s[1:])
    else:
        return s[0]+pairStar(s[1:])



print(pairStar('xxyy'))
print(pairStar('xaax'))
```

```python
#p8
def countAbc(s):
    if len(s)<=2:
        return 0
    if s[0]=='a':
        if s[1]=='b':
            if s[2]=='c' or 'a':
                return 1+countAbc(s[3:])
            else:
                return countAbc(s[2:])
        else:
            return countAbc(s[1:])
    else:
        return countAbc(s[1:])

print(countAbc('ababccabalkabc'))
```

```python
#p10
def strCount(ms,ss):
    if len(ms)<len(ss):
        return 0
    if ms[0:len(ss)]==ss:
        return 1+strCount(ms[len(ss):],ss)
    else:
        return strCount(ms[1:],ss)
    
print(strCount("cowatcowccow", "cow"))
```

```python
#p17
def parenBit(s):
    if s[0]=='*' and s[1]==')':
        return ')'
    if s[0]=='(':
        return s[0:2]+parenBit('*'+s[2:])
    elif s[0]=='*':
        return s[1]+parenBit('*'+s[2:])
    else:
        return parenBit(s[1:])

parenBit("(xyz)(abc)123")
```

```python
#p18
def strCopies(ms,ss,c,k = 0):

    if len(ms)<len(ss):
        if c<=0:
            return True
        else:
            return False
        
    else:
        if ms[0:len(ss):]==ss:
            return strCopies(ms[1:],ss,c-1)
        else:
            return strCopies(ms[1:],ss,c)

strCopies("catcowcat", "cow", 2)

```

```python
#p19
def changeXY(s):
    if len(s)==1:
        if s=='x':
            return 'y'
        else:
            return s
    else:
        if s[0]=='x':
            return 'y'+changeXY(s[1:])
        else:
            return s[0]+changeXY(s[1:])
        
changeXY("xxhixx")

```

```python
#p23
def stringClean(s):
    if len(s)==1:
        return s
    else:
        if s[0]!=s[1]:
            return s[0]+stringClean(s[1:])
        else:
            return stringClean(s[1:])
print(stringClean("hello"))
```

```python
#p24
def nestedParen(s):
    if len(s)==0:
        return True
    else:
        if s[0]=='(':
            if s[-1]==')':
                return nestedParen(s[1:-1])
            else:
                return False
        else:
            return False
print(nestedParen("((()))"))
```

```python
#p25
def strDist(ms,ss):
    
    if len(ms)<len(ss):
        return 0
    if ms[0:len(ss)]==ss:
        if ms[-1:-1-len(ss):-1][::-1]==ss:
            return len(ms)
        else:
            return strDist(ms[:-1:],ss)
    else:
        return strDist(ms[1::],ss)
    
print( strDist("catcowcat", "cow"))
```
